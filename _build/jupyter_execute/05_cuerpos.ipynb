```python
import random
from IPython.display import display, HTML

uid = str(random.randint(10000, 99999))

simulacion_html = """
<div style="border: 1px solid #e0e0e0; padding: 20px; border-radius: 8px; background-color: #f8f9fa; font-family: -apple-system, sans-serif;">
    <h3 style="margin-top:0; color: #2c3e50;">Simulación Avanzada: Sistema de Dos Masas (Atwood Modificada)</h3>
    
    <!-- Paneles de Solución (Construidos en HTML puro para compatibilidad en GitHub Pages) -->
    <div id="sol_sc1_UID" style="display: none; background: #e2e3e5; padding: 15px; border-radius: 5px; font-size: 0.95em; margin-bottom: 15px; color: #383d41;">
        <strong>Solución (Caso 1: Plano Liso - Sin Roce):</strong><br>
        Como la superficie es ideal (&mu; = 0), la fricción es nula (<i>f</i> = 0).
        <ul style="margin-top: 5px; margin-bottom: 5px;">
            <li>Bloque 1 (Eje X): <i>T = m<sub>1</sub> a</i></li>
            <li>Bloque 2 (Eje Y): <i>m<sub>2</sub> g - T' = m<sub>2</sub> a</i> (Considerando <i>T = T'</i>)</li>
        </ul>
        Sumando ambas ecuaciones para eliminar la tensión:
        <div style="text-align: center; font-size: 1.1em; margin: 10px 0;">
            <i>a = &nbsp;</i>
            <div style="display: inline-block; vertical-align: middle; text-align: center;">
                <div style="border-bottom: 1px solid #383d41; padding: 0 5px;"><i>m<sub>2</sub> g</i></div>
                <div style="padding: 0 5px;"><i>m<sub>1</sub> + m<sub>2</sub></i></div>
            </div>
        </div>
    </div>

    <div id="sol_sc2_UID" style="display: block; background: #e2e3e5; padding: 15px; border-radius: 5px; font-size: 0.95em; margin-bottom: 15px; color: #383d41;">
        <strong>Solución (Caso 2: Movimiento hacia la Derecha con Roce):</strong><br>
        La fricción cinemática (<i>f = &mu;<sub>k</sub> m<sub>1</sub> g</i>) apunta hacia la <strong>izquierda</strong>, oponiéndose al movimiento.
        <ul style="margin-top: 5px; margin-bottom: 5px;">
            <li>Bloque 1 (Eje X): <i>T - f = m<sub>1</sub> a</i></li>
            <li>Bloque 2 (Eje Y): <i>m<sub>2</sub> g - T' = m<sub>2</sub> a</i> (Considerando <i>T = T'</i>)</li>
        </ul>
        Sumando ambas ecuaciones para eliminar la tensión:
        <div style="text-align: center; font-size: 1.1em; margin: 10px 0;">
            <i>a = &nbsp;</i>
            <div style="display: inline-block; vertical-align: middle; text-align: center;">
                <div style="border-bottom: 1px solid #383d41; padding: 0 5px;"><i>m<sub>2</sub> g - &mu;<sub>k</sub> m<sub>1</sub> g</i></div>
                <div style="padding: 0 5px;"><i>m<sub>1</sub> + m<sub>2</sub></i></div>
            </div>
        </div>
    </div>

    <div id="sol_sc3_UID" style="display: none; background: #e2e3e5; padding: 15px; border-radius: 5px; font-size: 0.95em; margin-bottom: 15px; color: #383d41;">
        <strong>Solución (Caso 3: Movimiento hacia la Izquierda con Roce):</strong><br>
        Al empujar el bloque hacia la izquierda, la fricción (<i>f = &mu;<sub>k</sub> m<sub>1</sub> g</i>) se invierte y apunta hacia la <strong>derecha</strong> (misma dirección que <i>T</i>).
        <ul style="margin-top: 5px; margin-bottom: 5px;">
            <li>Bloque 1 (Eje X): <i>T + f = m<sub>1</sub> a</i></li>
            <li>Bloque 2 (Eje Y): <i>m<sub>2</sub> g - T' = m<sub>2</sub> a</i> (Considerando <i>T = T'</i>)</li>
        </ul>
        Sumando ambas ecuaciones se obtiene la aceleración que frena el sistema:
        <div style="text-align: center; font-size: 1.1em; margin: 10px 0;">
            <i>a = &nbsp;</i>
            <div style="display: inline-block; vertical-align: middle; text-align: center;">
                <div style="border-bottom: 1px solid #383d41; padding: 0 5px;"><i>m<sub>2</sub> g + &mu;<sub>k</sub> m<sub>1</sub> g</i></div>
                <div style="padding: 0 5px;"><i>m<sub>1</sub> + m<sub>2</sub></i></div>
            </div>
        </div>
    </div>
    
    <!-- Controles Superiores -->
    <div style="display: flex; gap: 20px; margin-bottom: 15px; background: #e9ecef; padding: 15px; border-radius: 8px; flex-wrap: wrap; align-items: center; justify-content: space-between;">
        
        <div style="display: flex; flex-direction: column; gap: 10px;">
            <label style="font-weight: 600; font-size: 0.85em; color: #495057;">Escenario de Estudio:</label>
            <select id="escenario_UID" style="padding: 8px; border-radius: 5px; border: 1px solid #ced4da; background: #fff; cursor: pointer; font-weight: bold;">
                <option value="1">1. Sin Roce (Plano Liso ideal)</option>
                <option value="2" selected>2. Con Roce (Inicia en Reposo)</option>
                <option value="3">3. Con Roce (Empujón a la Izquierda)</option>
            </select>
            
            <div style="display: flex; gap: 10px; margin-top: 5px;">
                <button id="play_UID" style="background-color: #198754; color: white; border: none; padding: 8px 15px; border-radius: 5px; cursor: pointer; font-weight: bold;">▶ Ejecutar</button>
                <button id="reset_btn_UID" style="background-color: #6c757d; color: white; border: none; padding: 8px 15px; border-radius: 5px; cursor: pointer; font-weight: bold;">🔄 Reiniciar</button>
            </div>
        </div>
        
        <div style="display: flex; gap: 20px; align-items: center;">
            <div style="display: flex; flex-direction: column; align-items: center; background: #fff; padding: 8px 15px; border-radius: 5px; border: 1px solid #ccc;">
                <label style="font-weight: 600; font-size: 0.85em; margin-bottom: 5px; color: #d63384;">Masa 1 (Horizontal)</label>
                <div style="display: flex; align-items: center; gap: 10px;">
                    <input type="range" id="m1_slider_UID" min="2" max="20" step="1" value="10" style="width: 120px;">
                    <span id="m1_val_UID" style="font-weight: bold; font-family: monospace; font-size: 0.9em; min-width: 45px;">10 kg</span>
                </div>
            </div>
            
            <div style="display: flex; flex-direction: column; align-items: center; background: #fff; padding: 8px 15px; border-radius: 5px; border: 1px solid #ccc;">
                <label style="font-weight: 600; font-size: 0.85em; margin-bottom: 5px; color: #198754;">Masa 2 (Colgante)</label>
                <div style="display: flex; align-items: center; gap: 10px;">
                    <input type="range" id="m2_slider_UID" min="2" max="20" step="1" value="12" style="width: 120px;">
                    <span id="m2_val_UID" style="font-weight: bold; font-family: monospace; font-size: 0.9em; min-width: 45px;">12 kg</span>
                </div>
            </div>
        </div>
    </div>
    
    <!-- Canvas de la Simulación -->
    <div style="display: flex; justify-content: center; margin-bottom: 15px;">
        <canvas id="canvas_sim_UID" width="850" height="320" style="background: #ffffff; border: 1px solid #ccc; border-radius: 4px; box-shadow: 0 2px 5px rgba(0,0,0,0.05); max-width: 100%;"></canvas>
    </div>
    
    <!-- Panel de Telemetría -->
    <div style="background: #ffffff; border: 1px solid #d6d8db; border-radius: 5px; padding: 15px; margin-bottom: 15px;">
        <div style="display: grid; grid-template-columns: repeat(auto-fit, minmax(110px, 1fr)); gap: 10px; text-align: center;">
            <div style="background: #f8f9fa; padding: 10px; border-radius: 5px; border: 1px solid #e9ecef;">
                <div style="font-size: 0.7em; color: #6c757d; font-weight: bold; text-transform: uppercase;">Estado del Sistema</div>
                <div id="lbl_estado_UID" style="font-size: 0.85em; font-weight: bold; color: #6c757d; margin-top: 5px; line-height: 1.2;">Reposo</div>
            </div>
            <div style="background: #e7f1ff; padding: 10px; border-radius: 5px; border: 1px solid #b6d4fe;">
                <div style="font-size: 0.7em; color: #084298; font-weight: bold; text-transform: uppercase;">Tensión (T)</div>
                <div id="lbl_T_UID" style="font-size: 1.1em; font-weight: bold; color: #0d6efd; margin-top: 5px; font-family: monospace;">0.0 N</div>
            </div>
            <div style="background: #f8d7da; padding: 10px; border-radius: 5px; border: 1px solid #f5c2c7;">
                <div style="font-size: 0.7em; color: #842029; font-weight: bold; text-transform: uppercase;">Fuerza Roce</div>
                <div id="lbl_fric_UID" style="font-size: 1.1em; font-weight: bold; color: #dc3545; margin-top: 5px; font-family: monospace;">0.0 N</div>
            </div>
            <div style="background: #fff3cd; padding: 10px; border-radius: 5px; border: 1px solid #ffecb5;">
                <div style="font-size: 0.7em; color: #997404; font-weight: bold; text-transform: uppercase;">Aceleración</div>
                <div id="lbl_a_UID" style="font-size: 1.1em; font-weight: bold; color: #fd7e14; margin-top: 5px; font-family: monospace;">0.00 m/s²</div>
            </div>
            <div style="background: #d1e7dd; padding: 10px; border-radius: 5px; border: 1px solid #badbcc;">
                <div style="font-size: 0.7em; color: #0f5132; font-weight: bold; text-transform: uppercase;">Velocidad</div>
                <div id="lbl_v_UID" style="font-size: 1.1em; font-weight: bold; color: #198754; margin-top: 5px; font-family: monospace;">0.00 m/s</div>
            </div>
        </div>
    </div>
    
    <!-- Filtros del DCL -->
    <div style="font-size: 0.95em; padding: 10px 15px; background: #e2e3e5; border-radius: 5px; color: #383d41; display: flex; flex-wrap: wrap; gap: 15px; justify-content: center; user-select: none;">
        <strong style="margin-right: 10px;">Vectores Visibles:</strong>
        <label style="cursor: pointer; display: flex; align-items: center; gap: 5px;">
            <input type="checkbox" id="chk_T_UID" checked style="cursor: pointer;">
            <span><b style="color:#0d6efd;">▬ T, T'</b> (Tensiones)</span>
        </label>
        <label style="cursor: pointer; display: flex; align-items: center; gap: 5px;">
            <input type="checkbox" id="chk_fric_UID" checked style="cursor: pointer;">
            <span><b style="color:#dc3545;">▬ f</b> (Roce)</span>
        </label>
        <label style="cursor: pointer; display: flex; align-items: center; gap: 5px;">
            <input type="checkbox" id="chk_P_UID" checked style="cursor: pointer;">
            <span><b style="color:#6f42c1;">▬ P₁, P₂, N</b> (Pesos y Normal)</span>
        </label>
        <label style="cursor: pointer; display: flex; align-items: center; gap: 5px;">
            <input type="checkbox" id="chk_A_UID" checked style="cursor: pointer;">
            <span><b style="color:#fd7e14;">▬ a</b> (Aceleración)</span>
        </label>
    </div>
</div>

<script>
    (function() {
        function initSimulation() {
            const canvas = document.getElementById('canvas_sim_UID');
            if (!canvas) { setTimeout(initSimulation, 50); return; }
            
            const ctx = canvas.getContext('2d');
            const btn_play = document.getElementById('play_UID');
            const btn_reset = document.getElementById('reset_btn_UID');
            const sel_escenario = document.getElementById('escenario_UID');
            
            const sol1 = document.getElementById('sol_sc1_UID');
            const sol2 = document.getElementById('sol_sc2_UID');
            const sol3 = document.getElementById('sol_sc3_UID');
            
            const m1_slider = document.getElementById('m1_slider_UID');
            const m2_slider = document.getElementById('m2_slider_UID');
            const m1_val = document.getElementById('m1_val_UID');
            const m2_val = document.getElementById('m2_val_UID');
            
            const lbl_T = document.getElementById('lbl_T_UID');
            const lbl_v = document.getElementById('lbl_v_UID');
            const lbl_fric = document.getElementById('lbl_fric_UID');
            const lbl_a = document.getElementById('lbl_a_UID');
            const lbl_estado = document.getElementById('lbl_estado_UID');
            
            const chk_T = document.getElementById('chk_T_UID');
            const chk_fric = document.getElementById('chk_fric_UID');
            const chk_P = document.getElementById('chk_P_UID');
            const chk_A = document.getElementById('chk_A_UID');
            
            let isPlaying = false;
            let finished = false; 
            let animationId;
            let lastTime = performance.now(); 
            
            const g = 10.0;        
            let mu_s = 0.5;      
            let mu_k = 0.4;      
            
            let m1 = 10.0;
            let m2 = 12.0;
            let P1 = 0, P2 = 0, N = 0;
            let T = 0, f_fric_mag = 0, f_fric_dir = -1, a = 0, v = 0;
            let s = 100.0; 
            
            let estado_txt = "Reposo";
            let color_estado = "#6c757d";

            function drawArrow(context, fromx, fromy, tox, toy, color, lineWidth=3, dashed=false) {
                const dx = tox - fromx;
                const dy = toy - fromy;
                const len = Math.hypot(dx, dy); 
                
                if (len < 0.5) return; 
                const headlen = Math.min(10, len * 0.5); 
                const angle = Math.atan2(dy, dx);
                
                context.beginPath();
                context.strokeStyle = color;
                context.lineWidth = lineWidth;
                if(dashed) context.setLineDash([4, 4]);
                context.moveTo(fromx, fromy);
                context.lineTo(tox, toy);
                context.stroke();
                context.setLineDash([]);
                
                context.beginPath();
                context.fillStyle = color;
                context.moveTo(tox, toy);
                context.lineTo(tox - headlen * Math.cos(angle - Math.PI / 6), toy - headlen * Math.sin(angle - Math.PI / 6));
                context.lineTo(tox - headlen * Math.cos(angle + Math.PI / 6), toy - headlen * Math.sin(angle + Math.PI / 6));
                context.fill();
            }

            function setupScenario() {
                const sc = sel_escenario.value;
                s = 100.0; 
                finished = false;
                
                sol1.style.display = (sc === "1") ? "block" : "none";
                sol2.style.display = (sc === "2") ? "block" : "none";
                sol3.style.display = (sc === "3") ? "block" : "none";
                
                if (sc === "1") {
                    mu_s = 0.0; mu_k = 0.0; v = 0.0;
                } else if (sc === "2") {
                    mu_s = 0.5; mu_k = 0.4; v = 0.0;
                } else if (sc === "3") {
                    mu_s = 0.5; mu_k = 0.4; v = -35.0; 
                }
                
                isPlaying = false;
                btn_play.innerText = "▶ Ejecutar";
                btn_play.style.backgroundColor = "#198754";
                btn_play.style.cursor = "pointer";
                
                updatePhysics(0);
                drawSim();
            }

            function updatePhysics(dt) {
                if (finished) return; 

                m1 = parseFloat(m1_slider.value);
                m2 = parseFloat(m2_slider.value);
                m1_val.innerText = m1.toFixed(1) + " kg";
                m2_val.innerText = m2.toFixed(1) + " kg";
                
                P1 = m1 * g;
                P2 = m2 * g;
                N = P1;
                
                const f_s_max = mu_s * N;
                const f_k = mu_k * N;
                
                // CORRECCIÓN DEL RETRASO INICIAL: Evaluamos en 'v === 0' estrictamente.
                if (v === 0) {
                    if (P2 <= f_s_max && sel_escenario.value === "2") {
                        a = 0;
                        f_fric_mag = P2; 
                        f_fric_dir = -1; 
                        T = P2;      
                        estado_txt = "Estático (Peso 2 insuficiente)";
                        color_estado = "#6c757d";
                    } else if (mu_s === 0 && sel_escenario.value === "1" && !isPlaying) {
                        // Reposo inicial para el escenario 1 antes de presionar Ejecutar
                        a = 0; f_fric_mag = 0; T = P2;
                        estado_txt = "En Reposo (Esperando ejecución)";
                        color_estado = "#6c757d";
                    } else {
                        // Inicia el movimiento inmediatamente sin pausa
                        f_fric_mag = f_k;
                        f_fric_dir = -1;
                        a = (P2 - f_k) / (m1 + m2);
                        T = m1 * a + f_k;
                        estado_txt = "Acelerando (Derecha)";
                        color_estado = "#dc3545";
                    }
                } else if (v > 0) {
                    f_fric_mag = f_k;
                    f_fric_dir = -1;
                    a = (P2 - f_k) / (m1 + m2);
                    T = m1 * a + f_k;
                    estado_txt = (a >= 0) ? "Acelerando (Derecha)" : "Frenando (Derecha)";
                    color_estado = (a >= 0) ? "#dc3545" : "#ffc107";
                } else if (v < 0) {
                    f_fric_mag = f_k;
                    f_fric_dir = 1; 
                    a = (P2 + f_k) / (m1 + m2); 
                    T = m1 * a - f_k; 
                    estado_txt = "Frenando (Mov. a la Izquierda)";
                    color_estado = "#0dcaf0";
                }
                
                v += a * dt;
                
                if (estado_txt === "Frenando (Mov. a la Izquierda)" && v > 0) {
                    v = 0; 
                }

                s += v * dt * 3; 
                
                if (s >= 250) { 
                    s = 250; v = 0; a = 0; finished = true; isPlaying = false;
                    estado_txt = "Polea Alcanzada"; color_estado = "#000";
                    btn_play.innerText = "🛑 Terminó"; btn_play.style.backgroundColor = "#dc3545";
                } else if (s <= 10) {
                    s = 10; v = 0; a = 0; finished = true; isPlaying = false;
                    estado_txt = "Tope Izquierdo Alcanzado"; color_estado = "#000";
                    btn_play.innerText = "🛑 Terminó"; btn_play.style.backgroundColor = "#dc3545";
                }
                
                lbl_T.innerText = T.toFixed(1) + " N";
                lbl_fric.innerText = (f_fric_mag > 0 && mu_k > 0 ? (f_fric_dir === 1 ? "+" : "-") : "") + f_fric_mag.toFixed(1) + " N";
                lbl_v.innerText = v.toFixed(1) + " m/s";
                lbl_a.innerText = a.toFixed(2) + " m/s²";
                lbl_estado.innerText = estado_txt;
                lbl_estado.style.color = color_estado;
            }

            function drawSim() {
                ctx.clearRect(0, 0, canvas.width, canvas.height);
                
                const tableY = 120;
                const pulleyX = 400;
                const pulleyR = 15;
                
                ctx.beginPath();
                ctx.strokeStyle = "#888";
                ctx.lineWidth = 4;
                ctx.moveTo(20, tableY);
                ctx.lineTo(pulleyX, tableY);
                ctx.stroke();
                
                ctx.beginPath();
                ctx.fillStyle = "#aaa";
                ctx.strokeStyle = "#555";
                ctx.lineWidth = 2;
                ctx.arc(pulleyX, tableY, pulleyR, 0, Math.PI*2);
                ctx.fill();
                ctx.stroke();
                ctx.fillStyle = "#333";
                ctx.arc(pulleyX, tableY, 3, 0, Math.PI*2);
                ctx.fill();
                
                const b1_x = 50 + s;
                const b1_y = tableY - 40; 
                const b2_x = pulleyX + pulleyR;
                const b2_y = tableY + s + 10;
                
                ctx.beginPath();
                ctx.strokeStyle = "#444";
                ctx.lineWidth = 2;
                
                ctx.moveTo(b1_x + 50, tableY - pulleyR); 
                ctx.lineTo(pulleyX, tableY - pulleyR);
                
                ctx.arc(pulleyX, tableY, pulleyR, -Math.PI/2, 0);
                
                ctx.lineTo(b2_x, b2_y);
                ctx.stroke();

                ctx.fillStyle = "#d63384"; 
                ctx.strokeStyle = "#a12663";
                ctx.fillRect(b1_x, b1_y, 50, 40);
                ctx.strokeRect(b1_x, b1_y, 50, 40);
                ctx.fillStyle = "white";
                ctx.font = "italic 16px serif";
                ctx.fillText("m₁", b1_x + 15, b1_y + 25);
                
                ctx.fillStyle = "#198754"; 
                ctx.strokeStyle = "#12653e";
                ctx.fillRect(b2_x - 20, b2_y, 40, 50);
                ctx.strokeRect(b2_x - 20, b2_y, 40, 50);
                ctx.fillStyle = "white";
                ctx.fillText("m₂", b2_x - 10, b2_y + 30);
                
                if (chk_A.checked && Math.abs(a) > 0.05) {
                    const asc = 12; 
                    drawArrow(ctx, b1_x + 25, b1_y - 20, b1_x + 25 + a * asc, b1_y - 20, "#fd7e14", 4);
                    drawArrow(ctx, b2_x + 30, b2_y + 25, b2_x + 30, b2_y + 25 + a * asc, "#fd7e14", 4);
                }

                ctx.beginPath();
                ctx.strokeStyle = "#e0e0e0";
                ctx.setLineDash([5, 5]);
                ctx.moveTo(480, 20);
                ctx.lineTo(480, canvas.height - 20);
                ctx.stroke();
                ctx.setLineDash([]);
                
                const sf = 0.55; 
                const dcl1_x = 600;
                const dcl1_y = 150;
                
                ctx.font = "bold 14px sans-serif";
                ctx.fillStyle = "#d63384";
                ctx.fillText("DCL: Bloque 1", dcl1_x - 45, 40);
                
                ctx.beginPath();
                ctx.strokeStyle = "#ddd";
                ctx.setLineDash([4, 4]);
                ctx.moveTo(dcl1_x - 70, dcl1_y); ctx.lineTo(dcl1_x + 70, dcl1_y);
                ctx.moveTo(dcl1_x, dcl1_y - 80); ctx.lineTo(dcl1_x, dcl1_y + 80);
                ctx.stroke();
                ctx.setLineDash([]);
                
                if (chk_P.checked) {
                    drawArrow(ctx, dcl1_x, dcl1_y, dcl1_x, dcl1_y - N * sf, "#212529", 3);
                    ctx.fillStyle = "#212529"; ctx.fillText("N", dcl1_x + 5, dcl1_y - N * sf + 10);
                    drawArrow(ctx, dcl1_x, dcl1_y, dcl1_x, dcl1_y + P1 * sf, "#6f42c1", 3);
                    ctx.fillStyle = "#6f42c1"; ctx.fillText("P", dcl1_x + 5, dcl1_y + P1 * sf - 5);
                    ctx.font = "bold 10px sans-serif";
                    ctx.fillText("1", dcl1_x + 5 + ctx.measureText("P").width, dcl1_y + P1 * sf - 1);
                }
                ctx.font = "bold 14px sans-serif"; // Reset font size
                if (chk_T.checked && T > 0) {
                    drawArrow(ctx, dcl1_x, dcl1_y, dcl1_x + T * sf, dcl1_y, "#0d6efd", 3);
                    ctx.fillStyle = "#0d6efd"; ctx.fillText("T", dcl1_x + T * sf - 15, dcl1_y - 10);
                }
                if (chk_fric.checked && f_fric_mag > 0) {
                    drawArrow(ctx, dcl1_x, dcl1_y, dcl1_x + (f_fric_mag * sf * f_fric_dir), dcl1_y, "#dc3545", 3);
                    ctx.fillStyle = "#dc3545"; ctx.fillText("f", dcl1_x + (f_fric_mag * sf * f_fric_dir), dcl1_y + 20);
                }

                const dcl2_x = 760;
                const dcl2_y = 150;
                
                ctx.fillStyle = "#198754";
                ctx.fillText("DCL: Bloque 2", dcl2_x - 45, 40);
                
                ctx.beginPath();
                ctx.strokeStyle = "#ddd";
                ctx.setLineDash([4, 4]);
                ctx.moveTo(dcl2_x, dcl2_y - 80); ctx.lineTo(dcl2_x, dcl2_y + 80);
                ctx.stroke();
                ctx.setLineDash([]);
                
                if (chk_T.checked && T > 0) {
                    drawArrow(ctx, dcl2_x, dcl2_y, dcl2_x, dcl2_y - T * sf, "#0d6efd", 3);
                    ctx.fillStyle = "#0d6efd"; ctx.fillText("T'", dcl2_x + 8, dcl2_y - T * sf + 10);
                }
                if (chk_P.checked) {
                    drawArrow(ctx, dcl2_x, dcl2_y, dcl2_x, dcl2_y + P2 * sf, "#6f42c1", 3);
                    ctx.fillStyle = "#6f42c1"; ctx.fillText("P", dcl2_x + 8, dcl2_y + P2 * sf - 5);
                    ctx.font = "bold 10px sans-serif";
                    ctx.fillText("2", dcl2_x + 8 + ctx.measureText("P").width, dcl2_y + P2 * sf - 1);
                }
                ctx.font = "bold 14px sans-serif"; // Reset font size
            }

            function animate(timestamp) {
                let dt = (timestamp - lastTime) / 1000;
                if (dt > 0.1) dt = 0.1; 
                lastTime = timestamp;
                
                if (isPlaying && !finished) {
                    updatePhysics(dt);
                }
                
                drawSim();
                animationId = requestAnimationFrame(animate);
            }

            m1_slider.addEventListener('input', () => { if(!isPlaying && !finished){ updatePhysics(0); drawSim(); } });
            m2_slider.addEventListener('input', () => { if(!isPlaying && !finished){ updatePhysics(0); drawSim(); } });
            sel_escenario.addEventListener('change', setupScenario);

            btn_play.addEventListener('click', () => {
                if (finished) return; 
                
                isPlaying = !isPlaying;
                if (isPlaying) {
                    btn_play.innerText = "⏸ Pausar";
                    btn_play.style.backgroundColor = "#ffc107";
                    btn_play.style.color = "#000";
                    lastTime = performance.now();
                } else {
                    btn_play.innerText = "▶ Ejecutar";
                    btn_play.style.backgroundColor = "#198754";
                    btn_play.style.color = "#fff";
                }
            });

            btn_reset.addEventListener('click', setupScenario);
            
            [chk_T, chk_fric, chk_P, chk_A].forEach(chk => {
                chk.addEventListener('change', () => {
                    if (!isPlaying) drawSim();
                });
            });
            
            setupScenario();
            requestAnimationFrame(animate);
        }
        
        initSimulation();
    })();
</script>
"""

display(HTML(simulacion_html.replace('UID', uid)))
```
